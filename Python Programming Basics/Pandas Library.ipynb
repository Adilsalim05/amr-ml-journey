####  Adding rows to the datafram 
The syntax df.loc[len(df)] = ... is a clever trick to add a row to the end of a DataFrame.Here is exactly how it works, broken down piece by piece:The Breakdownlen(df) counts the total number of rows.
- A 0-indexed DataFrame with 2 rows has index labels 0 and 1.For this DataFrame, len(df) returns exactly 2.
- df.loc[...] targets rows by their index label.df.loc[2] looks for a row labeled 2.
- Since index 2 does not exist yet, Pandas creates it.
- The assignment = ['Charlie', 35] fills this new row with your data.

```python
df.loc[len(df)] = ["Ahmad", 25, 88, True]
```

```python
df.loc[len(df)] = {
    "Name": "Jessica",
    "Age": 24,
    "Score": 95, 
    "Passed": True
}
df
```

## A Mental Model That Will Save You Time

- When using Pandas, ask yourself these four questions:

- How do I load the data? (read_csv, read_excel)
- How do I inspect it? (head, info, describe, shape)
- How do I select or transform it? (column selection, loc, iloc, filtering, apply)
- How do I summarize it? (groupby, value_counts, mean, describe)

- Most data analysis tasks follow this cycle repeatedly.

## A Practical Learning Path (Especially for Bioinformatics)

- Given your background in microbiology and bioinformatics, I suggest learning Pandas in this order:

    - Creating Series and DataFrame
    - Reading CSV/Excel files
    - Inspecting data (head, info, describe)
    - Selecting rows and columns (loc, iloc)
    - Filtering data
    - Handling missing values
    - Sorting and adding columns
    - Grouping and aggregation (groupby)
    - Merging and joining datasets
    - Working with dates and times
    - Reshaping data (pivot, melt)
    - Applying custom functions and optimizing performance

- Once you're comfortable with these, you'll be able to manipulate gene expression tables, sequencing metadata, antimicrobial susceptibility results, and other biological datasets efficiently.
