# merge ()  
- merge() is equivalent to Excel's VLOOKUP or XLOOKUP: it matches rows based on a shared key and brings over related information.

```python
advisors = pd.DataFrame({
    "Department": ["Biology", "Chemistry", "Physics"],
    "Advisor": ["Dr. Khan", "Dr. Smith", "Dr. Lee"]
})

advisors
```

```python
students.merge(advisors, on= "Department")
```

## The difference in one sentence

If you remember only one thing today, make it this:

groupby() = "Take one table and summarize it."
merge() = "Take two tables and combine them using a shared key."

### What does 943 mean?

It does not mean there are 943 different cephalosporin antibiotics.

It means:

There are 943 rows in the CARD database whose Drug Class field is exactly "cephalosporin".

Remember, groupby().size() counts rows, not unique drugs or genes.
