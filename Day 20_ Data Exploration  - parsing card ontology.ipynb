Why use entry.get()?

The roadmap uses:

entry.get("ARO_accession")

instead of

entry["ARO_accession"]

Why?

Imagine one entry doesn't have an accession.

With

entry["ARO_accession"]

Python raises:

KeyError

and your program stops.

With

entry.get("ARO_accession")

Python quietly returns:

None

and your loop continues.

This is much safer when working with real biological databases.

```python
import pandas as pd

records = []

for key, entry in card.items():
    if key.startswith("_") or not isinstance(entry, dict):
        continue
    records.append({
        "aro_accession": entry.get("ARO_accession"),
        "model_name": entry.get("model_name"),
        "model_type": entry.get("model_type"),
    })

card_df = pd.DataFrame(records)
```
